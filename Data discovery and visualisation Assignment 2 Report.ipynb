## 2.2 Data Cleansing <a id="part2.2"></a>

To clean my data, I
- Filled in my null values with 'NA'

```python
df['Air_Terminal'].fillna('NA',inplace = True)
df['Sea_Terminal'].fillna('NA',inplace = True)
df['Land_Terminal'].fillna('NA',inplace = True)
df['f4_industry'].fillna('NA',inplace = True)
df['f5_designation'].fillna('NA',inplace = True)
df['f5_designation.oth'].fillna('NA',inplace = True)
df['MainAccomm'].fillna('NA',inplace = True)
df['MainHotel'].fillna('NA',inplace = True)
df['travel_companion.2'].fillna('NA',inplace = True)
df['travel_companion.3'].fillna('NA',inplace = True)
df['travel_companion.4'].fillna('NA',inplace = True)
df['travel_companion.5'].fillna('NA',inplace = True)
```

- Fixed and standardised spelling

```python
naming_conventions = {'Accompaning':'Accompanying',
            'Administor':'Administer',
            'Attendent':'Attendant',
            'Dosmetic':'Domestic',
            'Enigneer':'Engineer',
            'Eingineer':'Engineer',
            'Entrepreuner':'Entrepreneur',
            'Enterpreneur':'Entrepreneur',
            'Hairdressor':'Hairdresser',
            'Journislist':'Journalist',
            'No stated':'Not Stated',
            'Partime':'Parttime',
            'Personel':'Personal',
            "Brother'S":"Brother's",
            'Sale girl':'Salesgirl',
            'Seamen':'Seaman',
            'Self employ':'Self-employed',
            'Self- employed':'Self-employed',
            'self-emplloyed':'Self-employed',
            'Self-employes':'Self-employed',
            'Self employed':'Self-employed',
            'SELF EMPLOYED':'Self-employed',
            'Fincial':'Financial',
            'Jourlist':'Journalist',
            'Mager':'Manager',
            'Persol':'Personal',
            'Professiols':'Professionals',
            'TRANSLATOR':'Translator',
            'Singapor':'Singapore'
             }
df.replace(naming_conventions,inplace = True)
```

- Capitalised the first letter for the column named f5_designation.oth

```python
df['f5_designation.oth'] = df['f5_designation.oth'].str.capitalize()
```

- Moderated the days to the an integer, took the average from the day ranges.

```python
days = {'1 day':1,
        '11-14 days':12.5,
        '15-29 days':22,
        '2 days':2,
        '3 days':3,
        '30-59 days':44.5,
        '4 days':4,
        '5 days':5,
        '6 days':6,
        '60 days & over':60,
        '7 days':7,
        '8-10 days':9,
        'Under 1 day':0.5}
df.replace(days,inplace = True)
```

# 3. Visualisations <a id="part3"></a>

## 3.1 Travellers Analysis <a id="part3.1"></a>

1. What percentage of the travellers are fluent in English?

```python
sns.set(rc={'figure.figsize':(14,10)})
fluency = df.groupby('langint')['case'].count()
fluency_type = fluency.index
fluency = fluency.sort_values(ascending=False)
fig, ax = plt.subplots()

ax.bar(fluency_type, fluency, color = 'm')
ax.set_title('Spoken language breakdown\n', fontsize = 20)
ax.set_ylabel('count', fontsize = 10)
ax.set_xlabel('language', fontsize=10)

for x, y in zip(fluency_type,fluency):
    label = y
    ax.annotate(label,
               (x,y),
               textcoords='offset points',
               xytext=(0,5),
               ha='center')
fig.tight_layout()
plt.show()
```

This visual show the breakdown of the different count of cases by spoken language. From this it shows the majority of Indonesians who travel to Singapore speaks Chinese and Malay is the least spoken language. Using this, stakeholders can identify the language the tours should offer.

2. What are the months with the highest number of travellers travelling to Singapore?

```python
import datetime

months = df['R.mth']
month_dt = []
for month in months:
    month_dt.append(datetime.datetime.strptime(month, '%B').month)
case = df.groupby(month_dt)['case'].count()
month = case.index.tolist()
month = sorted(month) 
case = case[month]

fig, axes = plt.subplots(nrows = 1, ncols = 1)

axes.plot(month, case, 'ko')
plt.plot(month,case,c='navy')

for x, y in zip(month,case): 

    label = y 

    axes.annotate(label, 
                 (x,y), 
                 textcoords="offset points", 
                 xytext=(0,10), 
                 ha='center') 

plt.title('Traveller count by month \n', fontsize = 20)
plt.ylabel('Count', fontsize = 10)
plt.xlabel('Months', fontsize=10)

plt.show()
```

This visual show the travellers case count by month. From this visual, it shows that March, August and December were the months with the most travellers visiting Singapore and Feburary, September and November were the months with the least travellers. Using this, stakeholders can identify which months to provide more tours and would require more resources.

3. How many of the travellers are revisting Singapore?

```python
color= ['#35978f','#80cdc1']
tourist = df[df['Purpose']=='Holiday/ Rest & Relax'].groupby('1st_visit')['case'].count()
sns.set(rc={'figure.figsize':(14,10)})
returns = tourist.index
fig2, ax = plt.subplots()
ax.pie(tourist,labels=returns, autopct='%1.2f%%', startangle=90, colors = color)
plt.Circle( (0,0), 0.7, color='white')
ax.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
ax.set_title('Reccurring tourists\n',fontsize=16)
#plt.show()
my_circle=plt.Circle( (0,0), 0.41, color='white')
p=plt.gcf()
p.gca().add_artist(my_circle)

plt.show()
```

This visual shows the percentage of recurring tourists. From this visual it shows the majority of the tourists here on holiday are here for the first time. Using this, stakeholders can come up with comprehensive tours that cover the top tourist attractions. For the recurring visitors we can provide a more indepth tours.

4. Between 2015 and 2016 are there more or less travellers coming to Singapore?

```python
color= ['#80cdc1','#35978f']
plt.pie(df[df['Purpose']=='Holiday/ Rest & Relax'].groupby('Year')['case'].count(), labels = df[df['Purpose']=='Holiday/ Rest & Relax']["Year"].value_counts().index,explode=(0.1,0.1), autopct='%1.1f%%',  shadow = True, colors = color)
ax.axis('equal')
plt.title('Between 2014 and 2015 which year has seen more travellers here on holiday\n',size = 14)
my_circle=plt.Circle( (0,0), 0.4, color='white')
p=plt.gcf()
p.gca().add_artist(my_circle)

plt.show()
```

From this visual it show the percentage of travellers in 2014 and 2015. This visual shows that the year 2014 had a higher percentage of travellers on holiday at 53.6%. From this, stakeholders can identify which year has more visitors.

5. What is the most popular accomodation type?

```python
df_holiday = df[df['Purpose'] == 'Holiday/ Rest & Relax']
companion1 = df_holiday['travel_companion.1'].value_counts()

df_companion2 = df[(df['Purpose'] == 'Holiday/ Rest & Relax') & (df['travel_companion.2'] != 'NA')]
companion2 = df_companion2['travel_companion.2'].value_counts()

df_companion3 = df[(df['Purpose'] == 'Holiday/ Rest & Relax') & (df['travel_companion.3'] != 'NA')]
companion3 = df_companion3['travel_companion.3'].value_counts()

df_companion4 = df[(df['Purpose'] == 'Holiday/ Rest & Relax') & (df['travel_companion.4'] != 'NA')]
companion4 = df_companion4['travel_companion.4'].value_counts()

df_companion5 = df[(df['Purpose'] == 'Holiday/ Rest & Relax') & (df['travel_companion.5'] != 'NA')]
companion5 = df_companion5['travel_companion.5'].value_counts()

companions = pd.concat([companion1, companion2, companion3, companion4, companion5], axis=0)
companions_df = companions.to_frame().reset_index()
companions_df.columns = ['Name', 'Count']
grouped = companions_df.groupby(['Name']).sum().reset_index()
grouped = grouped.sort_values(by='Count', ascending=False)
Accom = df[(df['Purpose'] == 'Holiday/ Rest & Relax') & (df['MainAccomm'] != 'NA')]
Accom_type = Accom['MainAccomm'].value_counts()
Accom_index = Accom_type.index
plt.xlim(0, 2150)
plt.title('Popularity of the different accomodation ', fontsize=20)
sns.barplot(x=Accom_type, y=Accom_index, data=grouped.reset_index(),palette="mako") 
plt.show()
```

From this visual, it shows the case count of the different accomodation types. This visual shows that hotels has the highest case count among indonesians on holiday and student hostel is the lowest. From this, stakeholders can identify that hotels are more popular and to offer hotel accomodation in their packaged tours.

## 3.2 Location Analysis <a id="part3.2"></a>

1. Where are travellers arriving from?

```python
sns.set_theme(style="white", context="talk")

df_clean1 = df[df['Air_Terminal'] != 'NA']
count_air = df_clean1['Air_Terminal'].value_counts()

df_clean2 = df[df['Sea_Terminal'] != 'NA']
count_sea = df_clean2['Sea_Terminal'].value_counts()

df_clean3 = df[df['Land_Terminal'] != 'NA']
count_land = df_clean3['Land_Terminal'].value_counts()

arrival_terminals = pd.concat([count_air,count_sea,count_land], axis=0)
palette = ['#0c2c84','#225ea8','#1d91c0','#41b6c4','#7fcdbb','#c7e9b4','#edf8b1','#ffffd9']
ax = sns.barplot(x=arrival_terminals.index, y=arrival_terminals, palette=palette,ci=None)

for i, v in enumerate(arrival_terminals):
    ax.text(i,v, str(v), color='black', fontweight='normal',fontsize = 14,ha='center', va='bottom')
plt.xticks(rotation=90)

ax.set_title('Travellers by terminal', fontsize=15)
ax.set_xlabel('Terminals')
ax.set_ylabel('Count of travellers')
plt.show()
```

This visual shows the count of travellers cases that arrive in Singapore from the different terminals. From this visual, the majority of the travellers arrived in Terminal 1 (Air terminal), HCC (Sea terminal) and Tuas (land terminal). From this visual, stakeholders can provide pick up services for those on day or packaged tours.

From this visual it shows the top 10 count of traveller cases by the city of residence. From this visual, it shows that the majority of traveller cases are from Jakarta which is the capital city of Indonesia with 1578 cases in the years of 2014 and 2015. From this visual, stakeholders can iddentify that Singapore is a popular holiday destination for the people of Jakarta and when our sales campaign should reach out to that city.

3. Which are the top 10 most popular hotel locations travellers stay in?

```python
mainhotel = df[(df["MainHotel"] != "NA") & (df['MainHotel'] != "Other Hotels")]
mainhotel = mainhotel["MainHotel"].value_counts()
top_5_hotels = mainhotel.sort_values(ascending=False)[:10]
sns.barplot(x=top_5_hotels.index, y=top_5_hotels,palette="cubehelix")
plt.xticks(rotation=90)
plt.title('Top 10 hotels with the highest count of cases\n', fontsize = 20)
plt.ylabel('count', fontsize = 10)
plt.xlabel('Hotel', fontsize=10)
plt.show()
```

From this visual, it show the top 10 hotels with the 10 highest case count. This visual shows that the top 3 hotels with the highest case count are Mandrin Orchard Singapore, V Hotel Lavender, York Hotel . From this, stakeholders can identify work with these hotels to promote our tours or work on a bundled hotel/tour package program.

```python
x_labels = ['totshopping_$','totacc_$','totfnb_$','tottran_$','totmedi_$','tototh_$','totbiz_$','totedu_$']
men_means = [df[(df['f1_gender']=='Male') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['totshopping_$']!='Na')]['totshopping_$'].mean(),
             df[(df['f1_gender']=='Male') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['totacc_$']!='Na')]['totacc_$'].mean(),
             df[(df['f1_gender']=='Male') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['totfnb_$']!='Na')]['totfnb_$'].mean(),
             df[(df['f1_gender']=='Male') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['tottran_$']!='Na')]['tottran_$'].mean(),
             df[(df['f1_gender']=='Male') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['totmedi_$']!='Na')]['totmedi_$'].mean(),
             df[(df['f1_gender']=='Male') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['tototh_$']!='Na')]['tototh_$'].mean(),
             df[(df['f1_gender']=='Male') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['totbiz_$']!='Na')]['totbiz_$'].mean(),
             df[(df['f1_gender']=='Male') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['totedu_$']!='Na')]['totedu_$'].mean()
             
              
            ]
women_means = [df[(df['f1_gender']=='Female') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['totshopping_$']!='Na')]['totshopping_$'].mean(),
               df[(df['f1_gender']=='Female') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['totacc_$']!='Na')]['totacc_$'].mean(),
               df[(df['f1_gender']=='Female') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['totfnb_$']!='Na')]['totfnb_$'].mean(),
               df[(df['f1_gender']=='Female') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['tottran_$']!='Na')]['tottran_$'].mean(),
               df[(df['f1_gender']=='Female') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['totmedi_$']!='Na')]['totmedi_$'].mean(),
               df[(df['f1_gender']=='Female') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['tototh_$']!='Na')]['tototh_$'].mean(),
               df[(df['f1_gender']=='Female') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['totbiz_$']!='Na')]['totbiz_$'].mean(),
               df[(df['f1_gender']=='Female') & (df['Purpose']=='Holiday/ Rest & Relax') & (df['totedu_$']!='Na')]['totedu_$'].mean()
               
               
              ]

x = np.arange(len(x_labels))  
width = 0.35

# Create a figure and subplot
fig, ax = plt.subplots()

# Plot side-by-side bars
bar1 = ax.bar(x - width/2, men_means, width, label='Men',color='b')
bar2 = ax.bar(x + width/2, women_means, width, label='Women',color='g')

# Customise the plot
ax.set_title('Average amount spent per product line by the different genders on holiday', fontsize = 24)
ax.set_ylabel('Amount($)', fontsize = 15)
ax.set_xticks(x)
ax.set_xticklabels(x_labels)
ax.set_ylim(bottom = 0, top = 300)
ax.legend()

# Create a function to display data labels/annotations for each bar
def autolabel(bars):
    for each in bars:
        height = each.get_height()
        ax.annotate('{:.1f}'.format(height),
                    xy=(each.get_x() + each.get_width() / 2, height),
                    xytext=(0, 2),  # 2 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom')

# Use function to display data labels for both bars
autolabel(bar1)
autolabel(bar2)

# Set tight layout for figure
fig.tight_layout()
plt.xticks(rotation=90)
# Display plot
plt.show()
```

This visual provides a breakdown of the spent by product line between the male and female Indonesians travellers in Singapore during their holiday.
